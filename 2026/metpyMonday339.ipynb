## 339 - Ways to Plot the Wind (streamlines, quiver, barbs)

[Youtube](https://www.youtube.com/watch?v=wskEe65-27U)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-4, 4, 25)
y = np.linspace(-4, 4, 25)
X, Y = np.meshgrid(x, y)
```

```python
# Uniform Flow
u_uniform = np.ones_like(X) * 8
v_uniform = np.ones_like(Y) * 2
```

```python
# Jet (speed gradient)
u_jet = 5 + 25 * np.exp(-(Y**2) / 1.5)
v_jet = np.zeros_like(X)
```

```python
# Curved Flow with Gradient
u_curved = 6 + 2.5 * X - 2.0 * Y
v_curved = 1.8 * X + 0.5 * Y
```

```python
def plot_wind_comparison(x, y, X, Y, u, v, title):
    speed = np.sqrt(u**2 + v**2)

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    # Streamlines with color by speed
    strm = axes[0].streamplot(x, y, u, v, color=speed)
    axes[0].set_title('Streamlines')

    # Quiver
    skip_quiver = (slice(None, None, 3), slice(None, None, 3))
    q = axes[1].quiver(X[skip_quiver], Y[skip_quiver], u[skip_quiver], v[skip_quiver], speed[skip_quiver])
    axes[1].set_title('Quiver')

    # Barbs
    skip_barbs = (slice(None, None, 3), slice(None, None, 3))
    axes[2].barbs(X[skip_barbs], Y[skip_barbs], u[skip_barbs], v[skip_barbs])
    axes[2].set_title('Barbs')

    fig.suptitle(title, fontsize=14)

```

```python
cases = [('Uniform Flow', u_uniform, v_uniform),('Jet-Like Flow', u_jet, v_jet), ('Curved Flow', u_curved, v_curved)]
```

```python
for title, u, v, in cases:
    plot_wind_comparison(x, y, X, Y, u, v, title)
```
