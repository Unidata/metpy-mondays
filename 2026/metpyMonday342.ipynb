```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import metpy.calc as mpcalc

from metpy.interpolate import cross_section
from metpy.units import units
from pathlib import Path
from urllib.request import urlretrieve
```

```python
date = '20230420'
hour = '00'
forecast_hour = '00'

filename = f'hrrr.t{hour}z.wrfprsf{forecast_hour}.grib2'

url = f'https://noaa-hrrr-bdp-pds.s3.amazonaws.com/hrrr.{date}/conus/{filename}'

Path(filename).parent.mkdir(parents=True, exist_ok=True)

urlretrieve(url, filename)
```

```python
ds = xr.open_dataset(filename, engine='cfgrib', filter_by_keys={'typeOfLevel': 'isobaricInhPa'})
```

```python
ds = ds.metpy.parse_cf()
```

```python
ds
```

```python
ds = ds.metpy.assign_crs({'grid_mapping_name': 'lambert_conformal_conic', 'latitude_of_projection_origin': 38.5, 'longitude_of_central_meridian': -97.5, 'standard_parallel': [38.5, 38.5], 'earth_radius': 6371229})

ds = ds.metpy.assign_y_x()
```

```python
ds
```

```python
start = (30, -97.2)
end = (45, -97.2)
```

```python
fig = plt.figure()
ax = plt.axes(projection = ccrs.PlateCarree())

ax.add_feature(cfeature.STATES)

ax.plot([start[1], end[1]], [start[0], end[0]], color = 'red', linewidth=2, marker='o', transform=ccrs.PlateCarree())
```

```python
cross = cross_section(ds, start, end)
```

```python
cross
```

```python
theta_cross = mpcalc.potential_temperature(cross['isobaricInhPa'], cross['t'])
```

```python
tangent_wind, normal_wind = mpcalc.cross_section_components(cross['u'], cross['v'])
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

rh_filled = ax.contourf(cross['index'], cross['isobaricInhPa'], cross['r'], levels = np.arange(40, 105, 5), cmap='YlGnBu', extend='both')

theta_contours = ax.contour(cross['index'], cross['isobaricInhPa'], theta_cross, levels=np.arange(280, 380, 4), colors='k', linewidths=1)
ax.clabel(theta_contours, fmt='%d')

step_x = 8
step_p = 2

ax.barbs(cross['index'][::step_x], cross['isobaricInhPa'][::step_p], tangent_wind[::step_p, ::step_x], normal_wind[::step_p, ::step_x], length=5)

ax.set_yscale('symlog')
ax.set_ylim(1000, 100)
ax.set_yticks(range(1000, 0, -100))
ax.set_ylabel('Pressure (hPa)')
ax.set_yticklabels([str(i) for i in range(1000, 0, -100)])

ax.set_xticks([])

cbar = fig.colorbar(rh_filled, ax=ax, pad=0.02)
cbar.set_label('Relative Humidity (%)')
```
